```python
documents = [
    "Machine Learning is a subset of artificial intelligence",
    "Deep learning uses neural networks with multiple layers",
    "Python is a programming language popular for data science",
    "India won the cricket world cup in 2011",
    "AI models can understand and generate human language"
]
```

```python
from  langchain_google_genai import GoogleGenerativeAIEmbeddings


embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2")  # gives 3072 dimensions for each document



vector = embeddings.embed_documents(documents)    # this is the manual way to convert text into embedding
```

```python
len(vector[0])
```

### Store vector Embeddings in Vector DB

```python
from langchain_chroma import Chroma
```

```python
vector_store = Chroma.from_texts(   # first automatically convert text to embeddings then store embeddings to vectorDB
    texts= documents,               # Chroma receives the original strings
    embedding=embeddings,           # Chroma calls the embedding model
    persist_directory= "./vector_db"   # Store the Chroma database files inside a folder called vector_db
)
```

### Fetch result from Vector DB

```python
query = "ai model vs human brain"

result = vector_store.similarity_search(query,k=2)  # k is no. of results required // by default 4 deta hai

result   # subse relevent info sbse uppr hogi
```
