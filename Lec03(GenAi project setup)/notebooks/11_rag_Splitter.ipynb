### Text splitter

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
texts = "Today, I want to speak from the heart about three pillars that define the future of our nation: our youth, our education system, and our farmers."
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 50 ,chunk_overlap = 20)
```

```python
chunks = splitter.split_text(texts)   # since we hv to split text so we used split_text
```

```python
chunks
```

### Document Text Loader

```python
from langchain_community.document_loaders  import TextLoader, PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
txt_ld = TextLoader("../data/speech.txt" , encoding="utf-8")
texts = txt_ld.load()
```

```python
pdf_ld = PyPDFLoader("../data/medical_report.pdf")
pdfs = pdf_ld.load()
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 1000, chunk_overlap = 200 )
```

```python
text_chunks = splitter.split_documents(texts)   # since we hv to split documents

text_chunks     # since hmne split document ko kiya hai to chunks bhi documents hi honge
# it is useful when we require a specific document
```

```python
pdf_chunks = splitter.split_documents(pdfs)
len(pdf_chunks)
```
