```python
from pydantic import BaseModel
from langgraph.graph import StateGraph,START,END
```

```python
class GreatState(BaseModel):     ## Global State
    message: str = ""
```

```python
graph = StateGraph(GreatState)   ## graph hv global state
```

## Node : it is a python function only

```python
def Greet(state:GreatState):   ## Node
    state.message = "Hello, how are you ?"
    return state
```

```python
graph.add_node("greet",Greet)   ## add this Node to graph
```

```python
graph.add_edge(START,"greet")
graph.add_edge("greet", END)
```
