```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langgraph.checkpoint.memory import MemorySaver
```

```python
llm = ChatGroq(model="openai/gpt-oss-20b")
search = GoogleSerperAPIWrapper()  ## created object of this
```

```python
search.run("latest dollor comparision with ruppee")  ## from this we can use google search
## but we want to build llm agent for the google search
```

#### now connect this tool to llm to build agent

```python
agent = create_agent(
    model= llm,
    tools = [search.run],
    system_prompt= "You are an agent and can search any question on google.",
    checkpointer= MemorySaver()   ## to maintain history
)
```

```python
question = "latest dollor comparision with ruppee"
```

```python
res = agent.invoke(
    {"messages": [{"role":"user","content":question}]},
    {"configurable":{"thread_id":"1"}}    ### for this id it will maintain seperate history, if id chanded then maintain seperate history
)
```
