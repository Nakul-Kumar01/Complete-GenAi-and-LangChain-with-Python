```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
from langchain.agents import create_agent
```

```python
@tool
def add_number(a:int ,b:int):
    #this is doc string
    """      
    it will return the sum of two numbers

    args:
        a : Number One
        b : Number Two
    """

    return a + b


@tool
def multiply_number(a:int ,b:int):
    #this is doc string
    """      
    it will return the product of two numbers

    args:
        a : Number One
        b : Number Two
    """

    return a * b




# add_number.invoke({"a":12,"b":8})
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

```python
agent = create_agent(  # agent is combination of these 3 things
    model=llm,
    tools = [add_number,multiply_number],
    system_prompt="You are a math teacher, and always use tool for calculation."
)
```

```python
res = agent.invoke({"messages":[{"role":"user","content":"what is sum of 2 and 3"}]})  ## This has covered T -> A -> O
# it is fixed that we hv to pass dict with fild "messages" , which will hv list of dict
```

```python
for x in res["messages"]:     ## give the whole process, but we want only the final result
    print(x,"\n")
res["messages"][-1].content[0]['text']  ## this will give the final answer
```
