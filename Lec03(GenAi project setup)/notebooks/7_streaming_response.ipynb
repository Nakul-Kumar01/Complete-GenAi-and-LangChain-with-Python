```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model = "gemini-3.5-flash-lite",streaming= True)
```

```python
query = "Can u explain me GenAi"
```

```python
# res ## this is generator , so we hv to iterate this generator
res = llm.stream(query)

for chunk in res:
    if chunk.content:
       print(chunk.content[0]["text"], end="")

    
```
