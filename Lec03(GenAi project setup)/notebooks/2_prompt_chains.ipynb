###  Dynamic Prompts :

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate  # for dynamic o/p
from langchain_core.output_parsers import StrOutputParser  ## for o/p parser

llm = ChatGoogleGenerativeAI(model = "gemini-3.5-flash-lite")
```

```python
prompts = ChatPromptTemplate.from_messages([
    {"role":"system", "content" : "u are {language} Translator"},
        {"role":"user", "content" : "{query}"},
])

print(prompts)
```

```python
final_prompt = prompts.format_messages(language = "hindi",query = "i love u")

print(final_prompt)  ### here it hv filled the arguments in intial prompt template
```

```python
res = llm.invoke(final_prompt)

print(res.content[0]["text"])
```

### Chains :

```python
prompts.invoke({"language":"hindi","query":"I love you"})   ### it is Runnable

## isska output shoud me input for llm.invoke(), usske liye we do:
res = llm.invoke(prompts.invoke({"language":"hindi","query":"I love you"}))  ## passing Dynamically
print(res.content[0]["text"])


##  But this is very long way
```

### Now using Chains :

```python
#  But Complete Chains is not this, now we will see it :
# prompts.invoke ka output, input hai llm.invoke ka
chains = prompts | llm   
res = chains.invoke({"language":"hindi","query":"I love you"})
print(res.content[0]["text"])
```

```python
## to get direct result we use out
out = StrOutputParser()    
chains = prompts | llm | out    ### this is like chain ## therefore it is langchain
res = chains.invoke({"language":"hindi","query":"I love you"})
print(res)   # now we don't need to use content and other things for clear o/p
```
