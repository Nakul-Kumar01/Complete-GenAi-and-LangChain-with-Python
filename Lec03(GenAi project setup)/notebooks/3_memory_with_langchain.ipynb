```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model = "gemini-3.5-flash-lite")
```

```python
res = llm.invoke("Hello, My name is Nakul")
res.content[0]["text"]
```

```python
res = llm.invoke("Hello, what is my name")
res.content[0]["text"]     ### llm does not storing our previous conversation details
```

###  Now we Maintain History of Our Converstion

```python
history = []
while True :
    query = input("User : ")
    if query.lower() in ["exit","quit","bye"]:
        print("good bye")
        break 

    history.append({"role":"user","content":query})
    print("User: ",query)


    res = llm.invoke(history)
    history.append({"role":"ai","content":res.content[0]["text"]})
    print("Ai: ",res.content[0]["text"],"\n")
```
