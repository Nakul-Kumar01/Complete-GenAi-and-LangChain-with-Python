###  Text Document Loader

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader(
    "../data/speech.txt",
    encoding="utf-8",
    )


texts = loader.load()
print(len(texts))
print(texts[0].page_content)
```

### PDF Document Loader

```python
from langchain_community.document_loaders import PyPDFLoader   ## using pypdf

loader = PyPDFLoader(
    "../data/medical_report.pdf",
)


docs = loader.load()
len(docs)   # gives 9 ,  since there is 9 pages in the document
```

```python
oneDoc = docs[0].page_content
print(oneDoc)
```

### CSV File Loader

```python
from langchain_community.document_loaders import CSVLoader
loader = CSVLoader("../data/employee_records.csv")


docs = loader.load()

print(len(docs))
docs[0].page_content
```

###  Web Page Loader

```python
from langchain_community.document_loaders import WebBaseLoader
loader = WebBaseLoader(web_path= "https://docs.langchain.com/oss/python/integrations/document_loaders")


docs = loader.load()

print(len(docs))
print(docs[0].page_content)
```

### Wikipedia Loader

```python
from langchain_community.document_loaders import WikipediaLoader

loader = WikipediaLoader(query = "Gen AI",load_max_docs=2)

docs = loader.load()         
```

```python
print(len(docs))      ## 2 because it loads 2 docs max
docs[0].page_content
```
